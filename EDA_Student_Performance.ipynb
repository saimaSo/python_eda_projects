```python
#outlier detection
#To detect outlier >> dist,boxplot,violinplot
```

```python
students_grades_df=df.copy()
students_grades_df
```

```python
#Q1-1.5*IQR
#Q3+1.5*IQR
```

```python
def detect_outliers(columns):
  for column in columns:
    Q1=students_grades_df[column].quantile(0.25)
    Q3=students_grades_df[column].quantile(0.75)
    IQR=Q3-Q1
    lower_fence=Q1-1.5*IQR
    upper_fence=Q3+1.5*IQR
    mask=(students_grades_df[column] >= lower_fence) & (students_grades_df[column] <= upper_fence)
    mask=mask.to_numpy()
    students_grades_df=students_grades_df[mask]
  return students_grades_df
```

```python
num_cols=['age','absences']
detect_outliers(num_cols)
```
