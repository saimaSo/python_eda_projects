```python
#data is either numerical or categorical , numerical can either be discrete and continuous
#discrete data though  is numerical but should be treated as categorical to do analysis
#But for model building all the categorical data should be converted to numbers
```

```python
cat_f=X.select_dtypes(include='object').columns
```

```python
num_f=X.select_dtypes(exclude='object').columns
```

```python
num_f
```

```python
#Scaling should be done with numerical featurs
#Categorical >>Onehot coding
```

```python
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.compose import ColumnTransformer #transforms the string col into int of pd data frame ,or array

#this estimator allows different columns or column subsets of the
#input to be transformed separately and the features generated by each
#transformer will be concatenated to form a single feature space
#This is useful for heterogenous or columnar data,to combine several
```

```python
numeric_transformer=StandardScaler()
oh_transformer=OneHotEncoder(drop='first')
preprocessor=ColumnTransformer([('OneHotEncoder',oh_transformer,cat_f),
 ('StandardScaler',numeric_transformer,num_f)])
```

```python
preprocessor
```

```python
X_train = preprocessor.fit_transform(Xtrain)
pd.DataFrame(X_train)
```
